```python


# --- Helper Functions ---

def truncate_coord(coord, decimals=4):
    """Truncates longitude and latitude to a specified number of decimal places."""
    factor = 10**decimals
    # Add small epsilon before int cast to handle potential floating point representation issues near the truncation boundary
    # e.g., 120.12349999999999 should become 120.1234
    epsilon = 1e-9
    return (int(coord[0] * factor + epsilon * math.copysign(1, coord[0])) / factor,
            int(coord[1] * factor + epsilon * math.copysign(1, coord[1])) / factor)

def is_on_segment_truncated(point, p1, p2, decimals=4):
    """
    Checks if a point lies on the segment [p1, p2] after truncating
    coordinates to 'decimals' places.
    """
    pt_trunc = truncate_coord(point, decimals)
    s1_trunc = truncate_coord(p1, decimals)
    s2_trunc = truncate_coord(p2, decimals)

    # If segment is just a point
    if s1_trunc == s2_trunc:
        return pt_trunc == s1_trunc

    # Check bounding box (inclusive)
    on_bbox = (min(s1_trunc[0], s2_trunc[0]) <= pt_trunc[0] <= max(s1_trunc[0], s2_trunc[0]) and
               min(s1_trunc[1], s2_trunc[1]) <= pt_trunc[1] <= max(s1_trunc[1], s2_trunc[1]))
    if not on_bbox:
        return False

    # Check for collinearity using cross-product on truncated coordinates
    # (y2 - y1) * (x3 - x2) - (y3 - y2) * (x2 - x1) == 0  (using s1, s2, pt)
    # Using truncated points: s1_trunc, s2_trunc, pt_trunc
    cross_product = ((s2_trunc[1] - s1_trunc[1]) * (pt_trunc[0] - s2_trunc[0]) -
                     (pt_trunc[1] - s2_trunc[1]) * (s2_trunc[0] - s1_trunc[0]))

    # Due to truncation, collinear points should ideally have a cross_product of exactly 0.
    # Use a small tolerance just in case of intermediate float representation nuances.
    is_collinear = math.isclose(cross_product, 0, abs_tol=1e-12)

    return is_collinear

def distance_sq(p1, p2):
    """Calculate squared Euclidean distance to avoid sqrt for comparison."""
    return (p1[0] - p2[0])**2 + (p1[1] - p2[1])**2

# --- Main Logic ---

# 1. Collect all unique points
all_points = set()
for road in roads:
    for coord in road.coords:
        all_points.add(coord)

print(f"Total unique points found: {len(all_points)}")

# 2. Find points to insert
# Structure: {road_index: {segment_start_index: [list_of_points_to_insert]}}
insertions_map = {}
modified_road_indices = set()

for point_to_check in all_points:
    point_truncated = truncate_coord(point_to_check, 4) # Pre-truncate for efficiency

    for road_idx, road in enumerate(roads):
        original_coords = list(road.coords)

        # Check if the exact point is already a vertex in this road
        is_original_vertex = False
        for vertex in original_coords:
             if vertex == point_to_check:
                 is_original_vertex = True
                 break
        if is_original_vertex:
            continue # Skip if this road already contains this exact point

        # Check if the point lies on any segment of this road (using truncated coords)
        for i in range(len(original_coords) - 1):
            p1 = original_coords[i]
            p2 = original_coords[i+1]

            if is_on_segment_truncated(point_to_check, p1, p2, 4):
                # Check again to ensure the truncated point isn't identical to truncated endpoints
                # This prevents adding points that are effectively duplicates after truncation
                p1_trunc = truncate_coord(p1, 4)
                p2_trunc = truncate_coord(p2, 4)
                if point_truncated != p1_trunc and point_truncated != p2_trunc:
                    # Store the original point_to_check, not the truncated one
                    segment_map = insertions_map.setdefault(road_idx, {})
                    point_list = segment_map.setdefault(i, [])
                    # Avoid adding the same point multiple times to the same segment list
                    if point_to_check not in point_list:
                        point_list.append(point_to_check)
                    modified_road_indices.add(road_idx)
                    # Found its place on this road, no need to check further segments
                    break # Go to the next road

print(f"Found potential insertions for {len(modified_road_indices)} roads.")

# 3. Reconstruct roads with insertions
new_roads = list(roads) # Start with a copy of the original roads

for road_idx in modified_road_indices:
    if road_idx not in insertions_map:
        continue # Should not happen based on logic, but safe check

    original_coords = list(roads[road_idx].coords)
    road_insertions = insertions_map[road_idx]
    new_coords_list = []

    for i in range(len(original_coords) - 1):
        p1 = original_coords[i]
        p2 = original_coords[i+1]

        # Add the starting point of the segment
        # Avoid adding consecutive duplicates
        if not new_coords_list or new_coords_list[-1] != p1:
            new_coords_list.append(p1)

        # Check if there are points to insert in this segment
        if i in road_insertions:
            points_to_insert = road_insertions[i]
            # Sort points based on distance from p1
            points_to_insert.sort(key=lambda p: distance_sq(p1, p))

            # Add the sorted points, avoiding duplicates with p1 or consecutive points
            for pt in points_to_insert:
                 if new_coords_list[-1] != pt:
                     new_coords_list.append(pt)

    # Add the very last point of the original LineString
    last_original_point = original_coords[-1]
    if not new_coords_list or new_coords_list[-1] != last_original_point:
        new_coords_list.append(last_original_point)

    # Final check for consecutive duplicates that might have slipped through
    final_coords = []
    if new_coords_list:
        final_coords.append(new_coords_list[0])
        for j in range(1, len(new_coords_list)):
            if new_coords_list[j] != new_coords_list[j-1]:
                final_coords.append(new_coords_list[j])

    # Create the new LineString if it has at least 2 points
    if len(final_coords) >= 2:
        new_roads[road_idx] = LineString(final_coords)
    else:
        # Handle cases where the process might result in a degenerate line
        print(f"Warning: Road {road_names[road_idx]} resulted in < 2 points after modification. Keeping original.")
        new_roads[road_idx] = roads[road_idx] # Keep original


# 4. Output the results

print("\n--- Modified Roads ---")
modified_road_names = []
for idx in sorted(list(modified_road_indices)):
    # Check if the road actually changed (comparing coordinate lists)
    if list(roads[idx].coords) != list(new_roads[idx].coords):
         modified_road_names.append(road_names[idx])

if not modified_road_names:
    print("No roads were modified.")
else:
    print("The following roads had points inserted:")
    for name in modified_road_names:
        print(f"- {name}")

print("\n--- Final Road Array ---")
output_strings = []
for i, road in enumerate(new_roads):
    coord_str = ", ".join([f"({lon}, {lat})" for lon, lat in road.coords])
    output_strings.append(f"    LineString([{coord_str}]), # {road_names[i]}")

print("roads = [")
print("\n".join(output_strings))
print("]")

# Example: Print details for a modified road (e.g., Road_2)
# check_idx = 1 # Index for Road_2
# if check_idx in modified_road_indices and list(roads[check_idx].coords) != list(new_roads[check_idx].coords):
#     print(f"\n--- Details for {road_names[check_idx]} ---")
#     print("Original Coords:", list(roads[check_idx].coords))
#     print("New Coords:", list(new_roads[check_idx].coords))
#     print("Insertions Found:", insertions_map.get(check_idx, {}))
```
